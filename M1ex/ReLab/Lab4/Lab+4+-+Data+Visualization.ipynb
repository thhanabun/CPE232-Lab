3. Calculate the average 'Attack' of each pokemon 'Type 1'. Which type has the most Attack stats? \
Ans: ___________

```python
# write your code here
atkBytype = df.groupby("Type 1").agg({"Attack":"mean"})
atkBytype.sort_values(by="Attack",ascending=False)[:1]
```

5. Create a visualization that show proportion of each pokemon Type 1 in the dataset\
**Hint:** plotly 

```python
# write your code here
plt.pie(df["Type 1"].value_counts(),labels=df["Type 1"].value_counts().index,autopct="%1.1f%%")
plt.show
```

6. Create a line plot of 'Attack', 'Defense', and 'HP' of Bulbasaur evolution. \
Bulbasaur -> Ivysaur -> Venusaur

```python
# write your code here
evoline = df.iloc[0:3]
evostat = evoline[["Name","Attack","Defense","HP"]]
plt.plot(evoline["Name"], evostat["Attack"])
plt.plot(evoline["Name"], evostat["Defense"])
plt.plot(evoline["Name"], evostat["HP"])
plt.xlabel("Generation")
plt.ylabel("Stat Value")
plt.show()
```
