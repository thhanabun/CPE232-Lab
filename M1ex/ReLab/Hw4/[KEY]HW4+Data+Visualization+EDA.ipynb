14. Create a USA State-Level Choropleth Map to visualize total sales per state.

*   The darkest color represents the highest total sales.
*   The lightest color represents the lowest total sales.
*   Use a continuous gradient scale (e.g., dark blue to light blue, dark red to light red, or any custom gradient of your choice).

*Hint: Use plotly.express*



```python
#Write your code here
import plotly.express as px

# Aggregate sales at the state level
state_sales = superstore_order.groupby('State')['Sales'].sum().reset_index()
state_sales.columns = ['State', 'Total Sales']

# Dictionary for state abbreviations
state_abbrev = {
    'Alabama': 'AL', 'Alaska': 'AK', 'Arizona': 'AZ', 'Arkansas': 'AR', 'California': 'CA',
    'Colorado': 'CO', 'Connecticut': 'CT', 'Delaware': 'DE', 'Florida': 'FL', 'Georgia': 'GA',
    'Hawaii': 'HI', 'Idaho': 'ID', 'Illinois': 'IL', 'Indiana': 'IN', 'Iowa': 'IA', 'Kansas': 'KS',
    'Kentucky': 'KY', 'Louisiana': 'LA', 'Maine': 'ME', 'Maryland': 'MD', 'Massachusetts': 'MA',
    'Michigan': 'MI', 'Minnesota': 'MN', 'Mississippi': 'MS', 'Missouri': 'MO', 'Montana': 'MT',
    'Nebraska': 'NE', 'Nevada': 'NV', 'New Hampshire': 'NH', 'New Jersey': 'NJ', 'New Mexico': 'NM',
    'New York': 'NY', 'North Carolina': 'NC', 'North Dakota': 'ND', 'Ohio': 'OH', 'Oklahoma': 'OK',
    'Oregon': 'OR', 'Pennsylvania': 'PA', 'Rhode Island': 'RI', 'South Carolina': 'SC', 'South Dakota': 'SD',
    'Tennessee': 'TN', 'Texas': 'TX', 'Utah': 'UT', 'Vermont': 'VT', 'Virginia': 'VA', 'Washington': 'WA',
    'West Virginia': 'WV', 'Wisconsin': 'WI', 'Wyoming': 'WY'
}

# Map state names to abbreviations
state_sales['State Code'] = state_sales['State'].map(state_abbrev)

# Create a USA state-level choropleth map
fig = px.choropleth(
    state_sales,
    locations='State Code',  # Use state abbreviations
    locationmode='USA-states',  # Ensure correct mapping
    color='Total Sales',  # The value to visualize
    color_continuous_scale="Blues",  # Gradient color scheme
    scope="usa",  # Focus only on USA
    labels={'Total Sales': 'Total Sales ($)'}
)

# Update the layout for better visualization
fig.update_layout(
    title_text="Total Sales Across U.S. States",
    geo=dict(showcoastlines=True)
)

# Show the map
fig.show()
```

15.2 Which shipping mode has the highest median profit?

Ans: Second Class

```python
import pandas as pd

# Calculate median profit for each shipping mode
median_profit = superstore_order.groupby("Ship Mode")["Profit"].median()

# Find the shipping mode with the highest median profit
highest_median_mode = median_profit.idxmax()
highest_median_value = median_profit.max()

# Display the results
print(f"The shipping mode with the highest median profit is: {highest_median_mode}")
print(f"Median profit for {highest_median_mode}: ${highest_median_value:.2f}")
```

[BONUS 20 pts] Determine the percentage of customers who:
- B1)returned the product once
- B2) returned the product at least once
- B3) never returned the product
- Finally, Plot a comparison of B2 and B3

*Note: please create additional cells to answer the above points*

```python
join = superstore_order.merge(superstore_return, on='Order ID', how='left')
print(join)
```

```python
join.groupby('Customer ID')['Returned'].count()
```

```python
Customer_once = join.groupby('Customer ID')['Returned'].count()[join.groupby('Customer ID')['Returned'].count() == 1].shape[0]
Customer_at_least_once = join.groupby('Customer ID')['Returned'].count()[join.groupby('Customer ID')['Returned'].count() >= 1].shape[0]
Customer_not_return = join.groupby('Customer ID')['Returned'].count()[join.groupby('Customer ID')['Returned'].count() <= 0].shape[0]

print(Customer_once, Customer_at_least_once, Customer_not_return)
```

```python
total = Customer_at_least_once + Customer_not_return

ratio = Customer_at_least_once / total * 100

print("returned percentage: ", ratio)
print("not returned percentage: ", 100 - ratio)
```

```python
#Plot a comparison of 10.2 and 10.3
plt.bar(['Returned', 'Not Returned'], [ratio, 100-ratio])
```
